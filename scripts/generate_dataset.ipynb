```python
import pandas as pd
import numpy as np
import random
from faker import Faker
import os

fake = Faker()
np.random.seed(42)
random.seed(42)

print("Generating dataset... this may take a few minutes.")

NUM_RECORDS = 3_000_000  # ~1GB

device_types = ['mobile', 'desktop', 'tablet']
browsers = ['Chrome', 'Firefox', 'Safari', 'Edge']
regions = ['North', 'South', 'East', 'West', 'Central']
membership = ['none', 'silver', 'gold', 'platinum']

chunk_size = 500_000
output_file = r'C:\Users\USER PC\bigdata_project\customer_behavior.csv'

for chunk_num in range(NUM_RECORDS // chunk_size):
    n = chunk_size
    df = pd.DataFrame({
        'customer_id': [fake.uuid4() for _ in range(n)],
        'age': np.random.randint(18, 70, n),
        'session_duration_mins': np.round(np.random.exponential(8, n), 2),
        'pages_visited': np.random.randint(1, 50, n),
        'clicks': np.random.randint(1, 200, n),
        'time_on_site_mins': np.round(np.random.exponential(12, n), 2),
        'previous_purchases': np.random.randint(0, 30, n),
        'cart_additions': np.random.randint(0, 15, n),
        'device_type': np.random.choice(device_types, n),
        'browser': np.random.choice(browsers, n),
        'region': np.random.choice(regions, n),
        'membership_level': np.random.choice(membership, n),
        'discount_applied': np.random.choice([0, 1], n, p=[0.7, 0.3]),
        'time_since_last_visit_days': np.random.randint(0, 365, n),
        'will_purchase': np.random.choice([0, 1], n, p=[0.65, 0.35])
    })

    if chunk_num == 0:
        df.to_csv(output_file, index=False, mode='w')
    else:
        df.to_csv(output_file, index=False, mode='a', header=False)

    print(f"  Written {(chunk_num+1) * chunk_size:,} records...")

print(f"\nDone! File saved to: {output_file}")
size_mb = os.path.getsize(output_file) / (1024 * 1024)
print(f"File size: {size_mb:.1f} MB")
```
